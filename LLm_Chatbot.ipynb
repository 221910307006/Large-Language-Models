```python
import torch
from transformers import pipeline
import gradio as gr
```

```python
model_name = "microsoft/phi-2"
```

```python
chatbot = pipeline(
    task="text-generation",
    model=model_name,
    torch_dtype=torch.float16,
    device_map="auto"
)
```

```python
def format_prompt(user_input):
    prompt = f"""
### Instruction:
You are a helpful AI assistant.

### User:
{user_input}

### Response:
"""
    return prompt
```

```python
def chat_response(user_input):
    prompt = format_prompt(user_input)

    output = chatbot(
        prompt,
        max_new_tokens=200,
        do_sample=True,
        temperature=0.7,
        top_p=0.9,
        repetition_penalty=1.1
    )

    return output[0]["generated_text"].split("### Response:")[-1].strip()
```

```python
interface = gr.Interface(
    fn=chat_response,
    inputs=gr.Textbox(lines=3, placeholder="Ask me anything..."),
    outputs=gr.Textbox(lines=10),
    title="🤖 LLM Chatbot (Phi-2)",
    description="Simple ChatGPT-like bot using Hugging Face pipeline"
)
```

```python
interface.launch(share=True)
```
